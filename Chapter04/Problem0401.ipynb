```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[1:4] >=0)
```

```julia
demand = [620, 490, 510, 380]
@constraint(model, .5 * demand .<= x .<= .7 *demand)
```

```julia
@constraint(model, sum(x) <= 1200)
```

```julia
@objective(model, Max, [1.60, 1.40, 1.90, 1.20] ⋅ x)
```

```julia
optimize!(model)
```

```julia
value.(x)
```

```julia
objective_value(model)
```
