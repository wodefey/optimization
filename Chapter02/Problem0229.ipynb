```julia
using JuMP
using HiGHS
using DataFrames
```

```julia
df = DataFrame(
    :Project => 1:8,
    :Profit => [2.1, 0.5, 3.0, 2.0, 1.0, 1.5, 0.6, 1.8],
    :Days => [550, 400, 300, 350, 450, 500, 350, 200],
    :CPU => [200, 150, 400, 450, 300, 150, 200, 600]
)

model = Model(HiGHS.Optimizer)

@variable(model, x[1:8], Bin)

@objective(model, Max, sum(df.Profit[i] * x[i] for i in 1:8))

@constraint(model, sum(df.CPU[i] * x[i] for i in 1:8) <= 1000)

@constraint(model, sum(x[i] for i in 1:8) >= 3)

@constraint(model, sum(df.Days[i] * x[i] for i in 1:8) <= 2400)

@constraint(model, sum(x[i] for i in [3, 4, 5, 8]) >= 1)
```

```julia
optimize!(model)
```

```julia
termination_status(model)
```
