```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[1:7] >= 0, Int)
```

```julia
@constraint(model, sum(x[i] for i ∈ 1:5) >= 8)
```

```julia
@constraint(model, sum(x[i] for i ∈ [1,2,3,4,7]) >= 6)
```

```julia
@constraint(model, sum(x[i] for i ∈ [1,2,3,6,7]) >= 6)
```

```julia
@constraint(model, sum(x[i] for i ∈ [1,2,5,6,7]) >= 6)
```

```julia
@constraint(model, sum(x[i] for i ∈ [1,4,5,6,7]) >= 6)
```

```julia
@constraint(model, sum(x[k] for k ∈ 3:7) >= 10)
```

```julia
@constraint(model, sum(x[k] for k ∈ 2:6) >= 10)
```

```julia
@objective(model, Min, sum(x[i] for i ∈ 1:7))
```

```julia
optimize!(model)
```
