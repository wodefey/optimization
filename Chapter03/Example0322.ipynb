```julia
using Plots
using ForwardDiff
```

```julia
x = 0.0:9.9
```

```julia
y1 = x -> (2.0 -0.3x) / 0.4
y2 = x -> (1.5 - 0.4x) / 0.2
y3 = x -> (0.5 - 0.2x) / 0.3
```

```julia
y_bound = min.(y1.(x), y2.(x))
```

```julia
plt = plot()


plot!(plt, x, y1.(x))
plot!(plt, x, y2.(x))
plot!(plt, x, y3.(x), fillrange=y_bound)
xlims!(plt, 0, 9.9)
ylims!(plt, 0, 6.0)
```

### Two Phase Improving Search

```julia
x2 = [0.0, 0.0, 2.0, 1.5, 0.5];
```

```julia
obj2 =  x -> [0.0, 0.0, 1.0, 1.0, 1.0] ⋅ x
```

```julia
grad_obj2 = x -> ForwardDiff.gradient(obj2, x)
```

```julia
grad_obj2(x2)
```
