```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[j=1:3, k=1:4] >= 0)
```

```julia
required = [70.0, 50.0, 85.0, 35.0]

@constraint(model, vec(sum(x, dims=1)) .>= required)
```

```julia
@constraint(model, sum(x, dims=2) .<= 80)
```

```julia
rating = [
    [90.0 80.0 10.0 50.0];
    [60.0 70.0 50.0 65.0];
    [70.0 40.0 80.0 85.0]
]

@objective(model, Max, rating ⋅ x)
```

```julia
optimize!(model)
```
