```julia
using JuMP
using HiGHS
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, β[1:2] >= 0)
@variable(model, sp[1:4] >= 0)
@variable(model, sm[1:4] >= 0)
@variable(model, d)
```

```julia
@constraint(model, β[1] + 2*β[2] - 1 == sp[1] - sm[1])
@constraint(model, β[1] + 3*β[2] - 3 == sp[2] - sm[2])
@constraint(model, β[1] + 5*β[2] - 3 == sp[3] - sm[3])
@constraint(model, β[1] + 7*β[2] - 5 == sp[4] - sm[4])
@constraint(model, sp[1] + sm[1] <= d)
@constraint(model, sp[2] + sm[2] <= d)
@constraint(model, sp[3] + sm[3] <= d)
@constraint(model, sp[4] + sm[4] <= d)
```

```julia
@objective(model, Min, d)
```

```julia
optimize!(model)
```

```julia
solution_summary(model)
```
