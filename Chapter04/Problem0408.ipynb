```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
candles = ["Santa", "tree", "house"]

modeling = [0.10, 0.10, 0.25]
decorating = [0.35, 0.15, 0.40]
packaging = [0.08, 0.03, 0.05]

profit = [16.0, 9.0, 27.0]
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[candles] >= 0.0, Int)
```

```julia
@constraint(model, modeling ⋅ x <= 20.0)
```

```julia
@constraint(model, decorating ⋅ x <= 60)
```

```julia
@constraint(model, packaging ⋅ x <= 20)
```

```julia
@objective(model, Max, profit ⋅ x)
```

```julia
optimize!(model)
```

```julia
value.(x)
```
