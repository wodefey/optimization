```julia
using JuMP
```

```julia
using Ipopt
```

```julia
model = Model(Ipopt.Optimizer)
```

```julia
@variable(model, l >= 0)
```

```julia
@variable(model, w >= 0)
```

```julia
@constraint(model, 2l + 2w <= 80)
```

```julia
@objective(model, Max, l*w)
```

```julia
optimize!(model)
```

```julia
value(l)
```

```julia
value(w)
```

```julia
objective_value(model)
```
