```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, b[1:4] >= 0, Int)
@variable(model, i[1:4] >= 0, Int)
```

```julia
@constraint(model, i[4] + b[1] == 2800 + i[1])
@constraint(model, i[1] + b[2] == 500 + i[2])
@constraint(model, i[2] + b[3] == 100 + i[3])
@constraint(model, i[3] + b[4] == 850 + i[4])
@constraint(model, b[1:4] .<= 1200)
```

```julia
@objective(model, Min, 15 * sum(i))
```

```julia
optimize!(model)
```

```julia
value.(b)
```

```julia
value.(i)
```

```julia
objective_value(model)
```
