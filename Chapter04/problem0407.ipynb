```julia
using JuMP
using LinearAlgebra
using HiGHS
```

### Let x be the cuttin pattern chosen

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[1:3] >= 0.0)
```

```julia
@constraint(model, 2x[1] + x[3] >= 37)
```

```julia
@constraint(model, 5x[2] + 3x[3] >= 211)
```

```julia
@objective(model, Min, (0.34x[1] + 0.22x[2] + .27x[3]))
```

```julia
optimize!(model)
```

```julia
value.(x)
```
