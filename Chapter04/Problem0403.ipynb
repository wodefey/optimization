```julia
using JuMP
using HiGHS
using LinearAlgebra
using DataFrames
```

```julia
feeds = ["Oat", "Corn", "Alfafa", "Hulls"]
cost = [200.0, 150.0, 100.0, 75.0];

content = DataFrame(
    :Oats => [0.60, 0.50, 0.90],
    :Corn => [0.80, 0.70, 0.30],
    :Alfafa => [0.55, 0.40, 0.60],
    :Hulls => [0.40, 1.00, 0.80]
);
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[feeds] >= 0)
```

```julia
@constraint(model,content[1,:] ⋅ x .>= .6)
@constraint(model,content[2,:] ⋅ x .<= .6)
@constraint(model,content[3,:] ⋅ x .>= .6
)
```

```julia
@constraint(model, sum(x) == 1.0)
```

```julia
@objective(model, Min, cost ⋅ x)
```

```julia
optimize!(model)
```
