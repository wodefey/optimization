```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
items = ["tableA", "tableB", "shortLeg", "longLeg", "top"]

@variable(model, x[items] >= 0, Int)
```

```julia
@constraint(model, x["shortLeg"] == 4x["tableA"])
```

```julia
@constraint(model, x["longLeg"] == 4x["tableB"])
```

```julia
@constraint(model, x["top"] == x["tableA"] + x["tableB"])
```

```julia
@constraint(model, 18x["shortLeg"] + 30x["longLeg"] <= 12*500)
```

```julia
@constraint(model, 0.3x["tableA"]+ 0.3x["tableB"] + 0.1x["shortLeg"] + 0.15x["longLeg"] + 0.5x["top"] <= 80.0)
```

```julia
@objective(model, Max, 30x["tableA"] + 45x["tableB"])
```

```julia
optimize!(model)
```
