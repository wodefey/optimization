```julia
using JuMP
using HiGHS
using Plots
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, 0 <= x <= 4)
```

```julia
@variable(model, z)
```

```julia
@constraint(model, z <= 2x + 3)
@constraint(model, z <= -x + 5)
```

```julia
@objective(model, Max, z)
```

```julia
optimize!(model)
```

```julia
objective_value(model)
```

```julia
value(x)
```

```julia
plt = plot(x -> min(2x + 3, -x + 5), 0, 4)
```
