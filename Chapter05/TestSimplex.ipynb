```julia
lm = LinearModel()
```

```julia
add_goal!(lm, MAXIMIZE)
```

```julia
lm
```

```julia
add_objective!(lm, [3.0, 4.0])
lm
```

```julia
add_constraint!(lm, Constraint([1.0, 3.0], LE, 7.0))
```

```julia
add_constraint!(lm, Constraint([2.0, 1.0], GE, 4.0))
lm
```

```julia
sm = build_simplex_model(lm)
sm
```

```julia
simplex(sm)
```

```julia
using JuMP
using HiGHS
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[1:2] >= 0)
```

```julia
@constraint(model,x[1] + 3x[2] <= 7)
```

```julia
@constraint(model,2x[1] + x[2] >= 4)
```

```julia
@objective(model, Max, 3x[1] + 4x[2])
```

```julia
optimize!(model)
objective_value(model)
value.(x)
```
