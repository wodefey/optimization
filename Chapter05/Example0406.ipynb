```julia
include("Simplex.jl")
```

```julia
lm = LinearModel()
add_goal!(lm,MINIMIZE)
add_objective!(lm, [1.0,1.0,1.0])
add_constraint!(lm, Constraint([1.0, 1.0, 1.0], GE, 10.0))
add_constraint!(lm, Constraint([0.0, 1.0, 1.0], GE, 7.0))
add_constraint!(lm, Constraint([1.0, 0.0, 1.0], GE, 7.0))
add_constraint!(lm, Constraint([1.0, 1.0, 0.0], GE, 7.0))
add_constraint!(lm, Constraint([1.0, 1.0, 1.0], GE, 9.0))
```

```julia
sm = optimize(lm)
sm
```
