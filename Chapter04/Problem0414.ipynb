```julia
using JuMP
using HiGHS
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, x[5:10] >= 0, Int)
```

```julia
@constraint(model, x[5] >= 2)
```

```julia
@constraint(model, sum(x[5:6]) >= 3)
```

```julia
@constraint(model, sum(x[5:7]) >= 5)
```

```julia
@constraint(model, sum(x[5:8]) >= 5)
```

```julia
@constraint(model, sum(x[6:9]) >= 3)
```

```julia
@constraint(model, sum(x[7:10]) >= 2)
```

```julia
@constraint(model, sum(x[8:10]) >= 4)
```

```julia
@constraint(model, sum(x[9:10]) >= 6)
```

```julia
@constraint(model, x[10] >= 3)
```

```julia
@objective(model, Min, sum(x))
```

```julia
optimize!(model)
```
