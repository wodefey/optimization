```julia
using JuMP
using HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, t[1:4] >= 0, Int)
@variable(model, i[1:3] >= 0, Int)
```

```julia
@constraint(model, t[1] == 100 + i[1])
@constraint(model, i[1] + t[2] == 130 + i[2])
@constraint(model, i[2] + t[3] == 95 + i[3])
@constraint(model, i[3] + t[4] == 300)
```

```julia
@objective(model, Min, [12,14,14, 14] ⋅ t + 1.2 * sum(i))
```

```julia
optimize!(model)
```

```julia
value.(t)
```

```julia
value.(i)
```
