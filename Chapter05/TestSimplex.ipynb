```julia
include("Simplex.jl")
using Test
```

### Testset
### Test 1: Application 5.1 Top Brass Trophy
### Test 2: Application 2.1Two Crude (Two-Phase)
### Test 3: Clever Clyde (Infeasible)

```julia
@testset "Simplex Model Tests" begin
    lm = LinearModel()
    add_goal!(lm, MAXIMIZE)
    add_objective!(lm, [12.0, 9.0])
    add_constraint!(lm, Constraint([1.0, 1.0], LE, 1750.0))
    add_constraint!(lm, Constraint([4.0, 2.0], LE, 4800.0))
    add_constraint!(lm, Constraint([1.0, 0.0], LE, 1400.0))
    add_constraint!(lm, Constraint([0.0, 1.0], LE, 1500.0))

    sm = optimize(lm)
    @test sm.termination_status == OPTIMAL
    @test sm.objective_value == 17700

    lm = LinearModel()
    add_goal!(lm, MINIMIZE)
    add_constraint!(lm, Constraint([0.3, 0.4], GE, 2.0))
    add_constraint!(lm, Constraint([0.4, 0.2], GE, 1.5))
    add_constraint!(lm, Constraint([0.2, 0.3], GE, 0.5))
    add_constraint!(lm, Constraint([1.0, 0.0], LE, 9.0))
    add_constraint!(lm, Constraint([0.0, 1.0], LE, 6.0))
    add_objective!(lm, [100.0, 75.0])
    
    sm = optimize(lm)
    @test sm.termination_status == OPTIMAL
    @test sm.objective_value == 462.5

    lm = LinearModel()
    add_goal!(lm, MAXIMIZE)
    add_objective!(lm, [1.0,1.0,1.0])
    add_constraint!(lm, Constraint([1.0, 1.0, 1.0], GE, 100.00))
    add_constraint!(lm, Constraint([0.0,0.0,1.0], LE, 5.0))
    add_constraint!(lm, Constraint([-0.5, 0.5, 0.5], EQ, 0.0))
    add_constraint!(lm, Constraint([0.4, -0.6, 0.4], EQ, 0.0))

    sm = optimize(lm)
    @test sm.termination_status == INFEASIBLE
    @test sm.two_phase == true
    @test sm.objective_value == 5.0
end
```

### LinearModel Error testing

```julia
@testset "Linear Model Error Tests" begin
    lm = LinearModel()

    @test_throws LinearModelError "No constraints specified in the linear model." optimize(lm)
    add_constraint!(lm, Constraint([1.0, 1.0], LE, 100.0))

    @test_throws LinearModelError "No objective specified in the linear model." optimize(lm)
end
```
