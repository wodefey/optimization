```julia
using JuMP
using HiGHS
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
items = [:std_pal, :long_pal, :std_sep, :long_sep, :cross_piece ]

@variable(model, x[items] >= 0, Int)
```

```julia
@constraint(model, x[:std_sep] == 3x[:std_pal])
@constraint(model, x[:long_sep] == 3x[:long_pal])
```

```julia
@constraint(model, x[:cross_piece] == 10x[:std_pal] + 18x[:long_pal])
```

```julia
@constraint(model, 0.25x[:std_pal] + 0.30x[:long_pal]  <= 200)
@constraint(model, 0.005x[:std_sep] + 0.007x[:long_sep] + 0.002x[:cross_piece] <= 40)
```

```julia
@objective(model, Max, 5.0x[:std_pal] + 7.0x[:long_pal])
```

```julia
optimize!(model)
```

```julia
value.(x)
```
