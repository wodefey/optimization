### Application 2.1 : Two Crude Petroleum

```julia
using JuMP
import HiGHS
using LinearAlgebra
```

```julia
model = Model(HiGHS.Optimizer)
```

```julia
@variable(model, 0 <= x <= 9)
```

```julia
@variable(model, 0 <= y <= 6)
```

```julia
@constraint(model, [0.3, 0.4] ⋅ [x, y] >= 2.0)
```

```julia
@constraint(model, [0.4, 0.2] ⋅ [x, y] >= 1.5)
```

```julia
@constraint(model, [0.2, 0.3] ⋅ [x, y] >= 0.5)
```

```julia
@objective(model, Min, [100.0, 75.0] ⋅ [x, y])
```

```julia
print(model)
```

```julia
optimize!(model)
```

```julia
round.(value.([x, y]), digits=2)
```

```julia
round(objective_value(model), digits=2)
```
