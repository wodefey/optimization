### Application 5.1 Top Brass Trophy

```julia
lm = LinearModel()
add_goal!(lm, MAXIMIZE)
add_objective!(lm, [12.0, 9.0])
add_constraint!(lm, Constraint([1.0, 1.0], LE, 1750.0))
add_constraint!(lm, Constraint([4.0, 2.0], LE, 4800.0))
add_constraint!(lm, Constraint([1.0, 0.0], LE, 1400.0))
add_constraint!(lm, Constraint([0.0, 1.0], LE, 1500.0))
lm
```

```julia
sm = optimize(lm)
sm
```

### Application 2.1Two Crude (Two-Phase)

```julia
lm = LinearModel()
add_goal!(lm, MINIMIZE)
add_constraint!(lm, Constraint([0.3, 0.4], GE, 2.0))
add_constraint!(lm, Constraint([0.4, 0.2], GE, 1.5))
add_constraint!(lm, Constraint([0.2, 0.3], GE, 0.5))
add_constraint!(lm, Constraint([1.0, 0.0], LE, 9.0))
add_constraint!(lm, Constraint([0.0, 1.0], LE, 6.0))
add_objective!(lm, [100.0, 75.0])
```

```julia
optimize(lm)
```
